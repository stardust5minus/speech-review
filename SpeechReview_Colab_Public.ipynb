## 1. 解包程序代码（无需另传项目 ZIP）

```python
import base64,hashlib,io,zipfile,sys
from pathlib import Path

_bundle = base64.b64decode("")
assert hashlib.sha256(_bundle).hexdigest() == "6554441dfb77cd16ad74dd14a40fa587b39554ed775d29a6263f3ab47a86b9e2", "程序包校验失败"
APP_ROOT = Path("/content/speech_review_code/6554441dfb77")
APP_ROOT.mkdir(parents=True,exist_ok=True)
with zipfile.ZipFile(io.BytesIO(_bundle)) as archive:
    for item in archive.infolist():
        destination = (APP_ROOT/item.filename).resolve()
        if APP_ROOT.resolve() not in destination.parents:
            raise ValueError("程序包路径越界")
    archive.extractall(APP_ROOT)
if str(APP_ROOT) not in sys.path:sys.path.insert(0,str(APP_ROOT))
print("程序已准备：",APP_ROOT)
```

## 2. 安装 Python 依赖
安装输出直接显示在本单元；不需要 VS Code 或本机 Python。

```python
#@title 安装工作台环境（自动修复 venv / pip）
import subprocess,sys,urllib.request
from pathlib import Path
VENV_ROOT=Path("/content/speech_review_python")
APP_PYTHON=str(VENV_ROOT/"bin/python")
def checked_step(command,log_path):
    print("Running:",command[1:3],"— please wait",flush=True)
    with Path(log_path).open("w",encoding="utf-8") as log:
        result=subprocess.run(command,stdout=log,stderr=subprocess.STDOUT)
    if result.returncode:
        print(Path(log_path).read_text(encoding="utf-8",errors="replace")[-10000:])
        raise RuntimeError(f"Setup failed. See {log_path}; copy the error above for troubleshooting.")
if not Path(APP_PYTHON).exists():
    # A custom Colab Python may lack ensurepip even after apt installs venv.
    # Try the active interpreter first; install system support only on failure.
    try:
        checked_step([sys.executable,"-m","venv","--without-pip",str(VENV_ROOT)],"/content/speech_review_venv.log")
    except RuntimeError:
        checked_step(["apt-get","update","-y"],"/content/speech_review_apt_update.log")
        checked_step(["apt-get","install","-y","python3-venv"],"/content/speech_review_apt_install.log")
        checked_step([sys.executable,"-m","venv","--without-pip",str(VENV_ROOT)],"/content/speech_review_venv.log")
pip_check=subprocess.run([APP_PYTHON,"-m","pip","--version"],stdout=subprocess.DEVNULL,stderr=subprocess.DEVNULL)
if pip_check.returncode:
    print("pip not found in venv. Bootstrapping pip...",flush=True)
    try:
        checked_step([APP_PYTHON,"-m","ensurepip","--default-pip"],"/content/speech_review_ensurepip.log")
    except RuntimeError:
        get_pip_path=Path("/content/speech_review_get_pip.py")
        try:
            urllib.request.urlretrieve("https://bootstrap.pypa.io/get-pip.py",get_pip_path)
            checked_step([APP_PYTHON,str(get_pip_path)],"/content/speech_review_get_pip.log")
        finally:
            get_pip_path.unlink(missing_ok=True)
    checked_step([APP_PYTHON,"-m","pip","--version"],"/content/speech_review_pip_check.log")
checked_step([APP_PYTHON,"-m","pip","install","--upgrade","pip"],"/content/speech_review_pip.log")
checked_step([APP_PYTHON,"-m","pip","install","-r",str(APP_ROOT/"requirements.txt")],"/content/speech_review_dependencies.log")
print("App dependencies installed in a separate environment; Colab's packages were not replaced.")
```

## 6. 启动后台与 Public URL
下载官方 cloudflared，生成随机密码并启动临时网址；不需要 Cloudflare 账号。重复运行会结束本单元上次启动的进程，数据保留。

```python
import json,time,urllib.request,urllib.error,secrets,base64,re,socket
PORT=8765
# Stop only processes created by this notebook, tunnel first.
for name in ("TUNNEL_PROCESS","REVIEW_PROCESS"):
    process=globals().get(name)
    if process is not None and process.poll() is None:
        process.terminate()
        try:process.wait(timeout=15)
        except subprocess.TimeoutExpired:process.kill();process.wait(timeout=10)
with socket.socket() as probe:
    if probe.connect_ex(("127.0.0.1",PORT))==0:
        raise RuntimeError("8765 端口被另一个服务占用，请先结束该服务或重启运行时。")
cloudflared=SESSION_ROOT/"cloudflared"
if not cloudflared.exists():
    temporary=cloudflared.with_suffix(".download")
    urllib.request.urlretrieve("https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64",temporary)
    temporary.chmod(0o700);temporary.replace(cloudflared)
WEB_USER="review"
WEB_PASSWORD=secrets.token_urlsafe(24)
AUTH_HEADER="Basic "+base64.b64encode(f"{WEB_USER}:{WEB_PASSWORD}".encode()).decode()
TUNNEL_LOG_PATH=SESSION_ROOT/"tunnel.log"
TUNNEL_LOG=TUNNEL_LOG_PATH.open("w",encoding="utf-8")
TUNNEL_PROCESS=subprocess.Popen([str(cloudflared),"tunnel","--url",f"http://127.0.0.1:{PORT}","--no-autoupdate","--protocol","http2"],stdout=TUNNEL_LOG,stderr=subprocess.STDOUT)
try:
    for attempt in range(60):
        matches=re.findall(r"https://[a-z0-9-]+\.trycloudflare\.com",TUNNEL_LOG_PATH.read_text(encoding="utf-8",errors="replace"))
        if matches:PUBLIC_URL=matches[-1];break
        if TUNNEL_PROCESS.poll() is not None:raise RuntimeError("隧道启动失败："+TUNNEL_LOG_PATH.read_text()[-2500:])
        time.sleep(1)
    else:raise RuntimeError("未获得 Public URL，请检查 tunnel.log 后重试。")
    env=os.environ.copy()
    env.update(SPEECH_REVIEW_PROXY_ORIGIN=PUBLIC_URL,SPEECH_REVIEW_WEB_USER=WEB_USER,SPEECH_REVIEW_WEB_PASSWORD=WEB_PASSWORD)
    REVIEW_LOG_PATH=SESSION_ROOT/"server.log"
    REVIEW_LOG=REVIEW_LOG_PATH.open("a",encoding="utf-8")
    REVIEW_PROCESS=subprocess.Popen([APP_PYTHON,"-m","uvicorn","backend.main:app","--host","127.0.0.1","--port",str(PORT)],cwd=APP_ROOT,env=env,stdout=REVIEW_LOG,stderr=subprocess.STDOUT)
    BASE=f"http://127.0.0.1:{PORT}/api"
    def api(path,payload=None):
        data=None if payload is None else json.dumps(payload,ensure_ascii=False).encode()
        request=urllib.request.Request(BASE+path,data=data,headers={"Content-Type":"application/json","Authorization":AUTH_HEADER})
        try:
            with urllib.request.urlopen(request,timeout=90) as response:return json.load(response)
        except urllib.error.HTTPError as exc:raise RuntimeError(exc.read().decode("utf-8",errors="replace")) from exc
    for attempt in range(60):
        try:api("/projects");break
        except Exception:
            if REVIEW_PROCESS.poll() is not None or attempt==59:
                raise RuntimeError("后台启动失败："+REVIEW_LOG_PATH.read_text(encoding="utf-8",errors="replace")[-4000:])
            time.sleep(1)
    print("后台已就绪，继续导入录音，然后打开第 8 步网址。")
except BaseException:
    for name in ("TUNNEL_PROCESS","REVIEW_PROCESS"):
        process=globals().get(name)
        if process is not None and process.poll() is None:process.terminate()
    raise
```

## 7. 载入录音：上传 ZIP 或填写 Drive 文件夹

ZIP 内放 WAV 与同名 TXT，可包含受试者子文件夹。不是上传本机文件夹路径（例如 `C:\Users\...`）；Colab 无法读取你电脑的磁盘。

如已挂载 Drive，可填写直接包含 WAV/TXT 的 `/content/drive/MyDrive/...` 路径，避免重新上传。默认不改变原始 WAV/TXT。自动识别 Number/Homo/Non-homo，也可手动指定。


```python
#@title 导入方式：在完整网页工作台直接导入
IMPORT_IN_WORKSPACE=True #@param {type:"boolean"}
if IMPORT_IN_WORKSPACE:
    print("继续第 8 步打开工作台，切换“对齐与边界编辑”，一键导入已有 WAV + TXT + TextGrid。")
    print("如果需要从 notebook 上传录音 ZIP，把 IMPORT_IN_WORKSPACE 改为 False 再运行本单元。")
else:
    #@title 导入受试者
    AUDIO_FOLDER = "" #@param {type:"string"}
    CONDITION = "unknown" #@param ["unknown","number","homo","nonhomo"]
    from colab_support import extract_inputs
    if AUDIO_FOLDER.strip():
        candidates=[Path(AUDIO_FOLDER.strip()).expanduser().resolve()]
        if not candidates[0].is_dir():raise ValueError("音频文件夹不存在，请检查 Drive 挂载和路径。")
    else:
        from google.colab import files
        uploaded=files.upload()
        names=[name for name in uploaded if name.lower().endswith(".zip")]
        if len(names)!=1:raise ValueError("每次请选择一个录音 ZIP 文件。")
        name=names[0]
        raw=uploaded[name]
        archive_key=hashlib.sha256(raw).hexdigest()[:16]
        archive_path=SESSION_ROOT/(archive_key+".zip")
        archive_path.write_bytes(raw)
        candidates=extract_inputs(archive_path,SAVE_ROOT/"inputs"/archive_key)
        del uploaded,raw
    if not candidates:raise ValueError("未找到 WAV 文件。")
    if len(candidates)==1:
        imported=api("/import",{"folder":str(candidates[0]),"condition":CONDITION})
        print("已导入：",imported["participant"],"，录音数：",len(imported["trials"]))
    else:
        print("检测到多个受试者，请在工作台中逐个复制以下目录并导入：")
        for folder in candidates:print(folder)
```

## 9. 可选：从 notebook 下载全部项目结果

如果 iPad 的 iframe 下载不方便，完成保存和对齐后，把下面开关改为 True 再运行。只导出已保存文本和有效对齐，自动／人工版本分开。不会把输入录音、模型、账号信息放进导出 ZIP。


```python
#@title 下载全部结果
DOWNLOAD_NOW = False #@param {type:"boolean"}
if DOWNLOAD_NOW:
    from google.colab import files
    projects=api("/projects")
    if not projects:raise ValueError("还没有导入项目。")
    result_path=SESSION_ROOT/"speech_review_results.zip"
    with zipfile.ZipFile(result_path,"w",zipfile.ZIP_DEFLATED) as archive:
        for project in projects:
            with urllib.request.urlopen(urllib.request.Request(BASE+f"/projects/{project['id']}/export-everything",headers={"Authorization":AUTH_HEADER}),timeout=120) as response:
                data=response.read()
            archive.writestr(f"{project['id']}/participant_export.zip",data)
    files.download(str(result_path))
else:
    print("尚未下载；需要时设 DOWNLOAD_NOW=True 并重新运行此单元。")
```
